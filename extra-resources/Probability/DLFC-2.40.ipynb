# Bent coin problem.

Let's say we have a bent coin, one side convex and another side convex. Let's also say that concave is more likely to face up (60% chance) and convex is likely to face up for 40% of the time when flipped.

Now, we don't know if the coin's head is convex or concave.

But we start with an assumption that chance of coin's head being convex is 0.1.

And we are told that among 10 flips, eight flips resulted in heads being up, and two flips resulted in tail being up.

Use Bayes' Theorem successively to arrive at final posterior probability that coin's head is convex.

## Notation and values

Let's first set the notations and values.

H : Coin's head is convex.

E : Head is up.


We know concave is more likely to face up (60%). So, our conditional probabilities become:

When hypothesis is true:

P(E | H) = P(Head faces up when head is convex ) = less likely = 0.4

P(¬E | H) = P(Head faces down when head is convex) = more likely = 0.6


When hypothesis is false:

P(E | ¬H) = P(Head faces up when head is concave) = more likely = 0.6

P(¬E | ¬H) = P(Head faces down when head is concave) = less likely = 0.4

So, in the experiments if:

1. Heads is up:

P(H | E) = (P(H) P(E | H)) / (P(H)P(E | H) + P(¬H)P(E | ¬H))

inserting values

P(H | E) = (P(H) * 0.4) / (P(H) * 0.4 + (1- P(H)) * 0.6)

2. Head is down

P(H | ¬E) = (P(H) P(¬E | H)) / (P(H) P(¬E | H) + P(¬H)P(¬E | ¬H))

inserting values

P(H | ¬E) = (P(H) * 0.6) / (P(H) * 0.6 + (1 - P(H)) * 0.4)

With each iteration our P(H) updates to reflect the evidences seen upto this point, it represents our up to date belief.


```python
prior = 0.1

for _ in range(6):
    # Event E:
    posterior = (0.4 * prior) / (0.4 * prior + 0.6 * (1-prior))
    prior = posterior
    print(prior)

for _ in range (2):
    # Event ¬E
    posterior = (0.6 * prior) / (0.6 * prior + 0.4 * (1-prior))
    prior = posterior
    print(prior)

for _ in range(2):
    # Event E:
    posterior = (0.4 * prior) / (0.4 * prior + 0.6 * (1-prior))
    prior = posterior
    print(prior)

```

```python
# doing the same thing other way:

def posterior_e(prior):
    return (0.4 * prior) / (0.4 * prior + 0.6 * (1-prior))
def posterior_not_e(prior):
    return (0.6 * prior) / (0.6 * prior + 0.4 * (1-prior))

experiments = [True] * 8 + [False] * 2 
prior = 0.1
print(f"Initial prior: {prior:.4f}")
print("Experiments: ", ["Head" if e else "Tail" for e in experiments])
for e in experiments:
    if e:
        prior = posterior_e(prior)
    else:
        prior = posterior_not_e(prior)
    print(f"""Posterior after {'E' if e else '¬E'}: {prior} (prior was {prior:.4f})""")
```
